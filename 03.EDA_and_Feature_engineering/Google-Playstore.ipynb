##### Thought Process: Cleaning the Size Column

`Step 1: Inspect the raw data first`

The Google Play Store Size column typically looks like this:
['19M', '14M', '8.7M', '25M', '2.8M', '512k', '3.0M', 'Varies with device', ...]

*Key observations from this inspection:*

Values are strings, not numbers.
Sizes are mixed units: M (Megabytes) and k (Kilobytes).
There's a non-numeric category: "Varies with device" — this isn't missing data (NaN), but it's not a usable number either. You need a strategy for it.

This inspection step is critical — you should never blindly write a conversion function before knowing what all the unique values look like. Skipping this is how you get silent bugs or NaN explosions later.

`Step 2: Decide on target unit`:

*Reasoning*: You need one consistent unit to make the column numeric and comparable. You have two choices:

Convert everything to `KB (kilobytes)` — avoids fractional values since M → K is multiplying by 1024 (or 1000).
Convert everything to `MB` — keeps numbers smaller/more interpretable, but K → M creates decimals (e.g., 512k → 0.5M).

Either is valid. I'll standardize to KB here since it avoids introducing decimals from the smaller unit, but the logic is symmetric if you prefer MB.


`Step 3: Decide what to do with "Varies with device"`

This is a feature engineering decision, not just cleaning:

Option A: Convert to`NaN` and impute later (e.g., median size, or median size grouped by Category).
Option B: Create a separate binary flag column (Size_Varies = 1) before nulling it out — because "varies with device" might itself carry signal (e.g., apps with device-dependent size might behave differently, like games vs. utility apps).

I'd recommend Option B — it preserves information rather than discarding it.`

`Step 4: Write the conversion function`

```python
def convert_size(size):
    if size == 'Varies with device':
        return np.nan
    elif size.endswith('M'):
        return float(size[:-1]) * 1024   # MB -> KB
    elif size.endswith('k'):
        return float(size[:-1])          # already in KB
    else:
        return np.nan  # catch any unexpected format
```

`Why write it as a function with explicit branches instead of a one-liner regex replace?`

It's easier to read and debug.
The else: return np.nan branch acts as a safety net — if the dataset has some unexpected format you didn't anticipate during Step 1, it won't crash your pipeline; it'll just surface as a NaN you can investigate.
