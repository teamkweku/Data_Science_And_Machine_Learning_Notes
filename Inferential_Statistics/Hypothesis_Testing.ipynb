# Hypothesis Testing — Starting From Zero

---

## The Simplest Possible Analogy First

Forget statistics for a moment.

A pharmaceutical company comes to you and says:

> *"Our new drug reduces blood pressure. Trust us."*

You say: *"Prove it."*

That back-and-forth — that demand for proof — **is hypothesis testing.** It is nothing more than a formal framework for deciding whether a claim about the world is supported by evidence, or whether what you observed could have just been random chance.

The entire machinery of hypothesis testing exists to answer one question:

> **"Is what I'm seeing real — or could it have happened by luck?"**

---

## The Courtroom Analogy

This is the most intuitive way to understand the logic.

In a courtroom, the defendant starts **innocent until proven guilty.** The prosecution does not need to prove innocence — they need to provide enough evidence to establish guilt *beyond reasonable doubt.*

Hypothesis testing works the same way:

- The **null hypothesis (H₀)** is innocence — the default position, the boring explanation. *"Nothing is happening. The drug doesn't work. The difference is just random noise."*

- The **alternative hypothesis (H₁)** is the prosecution's claim — *"Something real is happening. The drug works. This difference is too large to be coincidence."*

- Your **data is the evidence** presented in court

- The **p-value** is how convincing the evidence is — specifically, how likely you'd see evidence this strong if the defendant were actually innocent

- **Rejecting H₀** is the verdict of guilty — you have enough evidence to conclude something real is happening

Just like a court never *proves* innocence — it only *fails to convict* — statistics never proves the null hypothesis. It either rejects it, or fails to reject it.

---

## What Is a Z-Test Specifically

Now that you have the logic, the Z-test is simply one specific tool within that framework.

You use a Z-test when you want to ask:

> *"Is the average of my sample meaningfully different from a known population average?"*

**The clinical scenario:**

The national average systolic blood pressure for adults in Ghana is known to be **128 mmHg.** You work at a hospital in Accra and you measure BP in a sample of 40 of your patients. Their average comes out at **135 mmHg.**

Your question: *Is my hospital's patient panel genuinely different from the national average — or could this 7 mmHg gap just be random sampling variation?*

That is a Z-test question.

---

## The Z-Score — What It Actually Means

Before the test, you need to understand the Z-score intuitively.

A Z-score answers: **"How many standard deviations away from the expected value is my result?"**

Think of it like this. Imagine the national BP average is 128, and from historical data you know that if you repeatedly sampled 40 random Ghanaian adults, the average of those samples would vary with a standard deviation of about 3 mmHg — just due to random chance.

Your sample gave you 135. That is 7 mmHg above the expected 128.

In units of that natural sampling variation (3 mmHg):

$$Z = \frac{135 - 128}{3} = 2.33$$

Your result sits **2.33 standard deviations above** what you'd expect by chance.

Now ask yourself intuitively — if most random samples land within 1–2 standard deviations of the expected value, a result sitting 2.33 standard deviations away is starting to look suspicious. It *could* be chance — but it's getting unlikely.

The formal Z-score formula is:

$$Z = \frac{\bar{x} - \mu_0}{\sigma / \sqrt{n}}$$

Where:
- $\bar{x}$ = your sample mean
- $\mu_0$ = the known population mean you're comparing against
- $\sigma$ = the known population standard deviation
- $n$ = your sample size
- $\sigma / \sqrt{n}$ = the **standard error** — how much sample means naturally vary just by chance

The standard error is the denominator doing the crucial work — it accounts for the fact that larger samples produce more reliable averages, so the same gap means more with n=200 than with n=10.

---

## The P-Value — What It Actually Means

The p-value is the most misunderstood number in all of statistics. Let's fix that right now with a plain definition:

> **The p-value is the probability of seeing a result as extreme as yours — or more extreme — if the null hypothesis were actually true.**

Back to the courtroom: if your patient panel's average BP is *actually* no different from the national average (H₀ is true), how often would random sampling alone produce a gap of 7 mmHg or larger?

If that probability is very small — say 2% — then either:
- You got very unlucky with your random sample, OR
- The null hypothesis is wrong and your patients really are different

The smaller the p-value, the harder it is to believe the null hypothesis.

**The decision threshold — alpha (α):**

By convention, researchers pre-agree on a threshold — almost always **α = 0.05** — before seeing the data. This means: *"I will consider the evidence convincing if there is less than a 5% chance of seeing this result under the null hypothesis."*

- **p-value < 0.05** → Reject H₀ → *"The difference is statistically significant"*
- **p-value ≥ 0.05** → Fail to reject H₀ → *"Not enough evidence to conclude a real difference"*

---

## Now Let's Walk a Real Problem From Start to Finish

**The question:**

> A medical researcher claims that the average fasting blood glucose of patients attending a diabetes clinic in Accra is higher than the national average of **92 mg/dL.** She collects a sample of **36 patients** from the clinic. Their mean fasting glucose is **98 mg/dL.** From national health records, the population standard deviation is known to be **18 mg/dL.** Is there sufficient evidence at the 5% significance level to support her claim?

---

### Step 1 — Read the question and identify every number

This is the first thing you do before writing a single formula. Just extract:

| What | Value |
|---|---|
| Known population mean (μ₀) | 92 mg/dL |
| Sample mean (x̄) | 98 mg/dL |
| Population standard deviation (σ) | 18 mg/dL |
| Sample size (n) | 36 patients |
| Significance level (α) | 0.05 |

---

### Step 2 — State your hypotheses in plain English first, then mathematically

**Ask yourself:** What is the researcher claiming? She says glucose is *higher* than national average — not just *different*, but specifically *higher.* That word matters.

**In plain English:**
- H₀: *"The clinic's patients have the same average glucose as the national average — any difference is just chance"*
- H₁: *"The clinic's patients genuinely have higher average glucose than the national average"*

**Mathematically:**
$$H_0: \mu = 92 \text{ mg/dL}$$
$$H_1: \mu > 92 \text{ mg/dL}$$

**Notice the direction:** H₁ uses "greater than" not "not equal to." This makes it a **one-tailed test** — you are only looking for evidence in one direction (higher). If she had simply said *"different from"* without specifying direction, it would be a **two-tailed test** using ≠.

---

### Step 3 — Calculate the Standard Error

The standard error tells you how much sample means naturally vary just by random chance for samples of this size.

$$SE = \frac{\sigma}{\sqrt{n}} = \frac{18}{\sqrt{36}} = \frac{18}{6} = 3 \text{ mg/dL}$$

**Intuition check:** With samples of 36 patients, random sampling alone would make the average bounce around by about ±3 mg/dL. Keep that in mind.

---

### Step 4 — Calculate the Z-Score

$$Z = \frac{\bar{x} - \mu_0}{SE} = \frac{98 - 92}{3} = \frac{6}{3} = 2.0$$

**Intuition check:** Your sample mean sits exactly **2 standard errors above** the national average. Is 2 standard errors a lot? Your instinct should be saying *"that's starting to feel unlikely by pure chance"* — and you'd be right.

---

### Step 5 — Find the P-Value

A Z-score of 2.0 on the standard normal distribution corresponds to a p-value of **0.0228** for a one-tailed test.

You don't need to memorise this — you look it up in a Z-table or let Python calculate it. But understand what it means:

> *"If the true population mean really were 92 mg/dL, there is only a 2.28% chance of randomly drawing a sample of 36 patients with a mean of 98 mg/dL or higher."*

---

### Step 6 — Make the Decision

Compare your p-value to your significance level:

$$p\text{-value} = 0.0228 \quad \text{vs} \quad \alpha = 0.05$$

$$0.0228 < 0.05 \quad \checkmark$$

**Decision: Reject H₀**

---

### Step 7 — State Your Conclusion in Plain Clinical Language

> *"At the 5% significance level, there is sufficient evidence to conclude that the mean fasting blood glucose of patients attending this Accra diabetes clinic is significantly higher than the national average of 92 mg/dL (Z = 2.0, p = 0.023)."*

Notice the conclusion always: states the decision, references the significance level, translates back to clinical meaning, and reports Z and p together.

---

## Now Let's See This in Python---

## The Seven Questions to Ask Every Time

Stick this on the wall above your desk. Every hypothesis testing problem — Z-test, t-test, chi-square, ANOVA — starts with these same seven questions:

| # | Question | In This Problem |
|---|---|---|
| 1 | What is the claim being tested? | Clinic glucose is higher than national average |
| 2 | What are my H₀ and H₁? | H₀: μ=92, H₁: μ>92 |
| 3 | One-tailed or two-tailed? | One-tailed — claim specifies a direction |
| 4 | What are my known values? | μ₀=92, x̄=98, σ=18, n=36 |
| 5 | What is my standard error? | SE = 18/√36 = 3 |
| 6 | What is my test statistic and p-value? | Z=2.0, p=0.023 |
| 7 | What is my conclusion in plain language? | Sufficient evidence — reject H₀ |

---

## One Thing to Sit With

The p-value does **not** tell you the probability that H₀ is true. It tells you the probability of your *data* given that H₀ is true. That is a subtle but important distinction that trips up even experienced researchers.

A p-value of 0.023 does not mean there is a 2.3% chance the national average is correct. It means: *if the national average were correct, you'd only see a result this extreme 2.3% of the time.* Those are different statements.

---

### Now Let's See This in Python

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

sns.set_theme(style="whitegrid", font_scale=1.1)

# ── Step 1: Define the known values ───────────────────────────────
mu_0   = 92      # known population mean (national average)
x_bar  = 98      # sample mean
sigma  = 18      # known population standard deviation
n      = 36      # sample size
alpha  = 0.05    # significance level

# ── Step 2: Calculate standard error and Z-score ──────────────────
SE = sigma / np.sqrt(n)
Z  = (x_bar - mu_0) / SE

# ── Step 3: Calculate p-value (one-tailed — testing if HIGHER) ────
p_value = 1 - stats.norm.cdf(Z)    # area to the RIGHT of Z

# ── Step 4: Decision ──────────────────────────────────────────────
decision = "REJECT H₀" if p_value < alpha else "FAIL TO REJECT H₀"

print("Z-Test Results — Diabetes Clinic Glucose Study")
print(f"{'─'*46}")
print(f"  Population mean (μ₀)  : {mu_0} mg/dL")
print(f"  Sample mean (x̄)       : {x_bar} mg/dL")
print(f"  Population SD (σ)     : {sigma} mg/dL")
print(f"  Sample size (n)       : {n}")
print(f"{'─'*46}")
print(f"  Standard Error (SE)   : {SE:.3f} mg/dL")
print(f"  Z-score               : {Z:.3f}")
print(f"  P-value (one-tailed)  : {p_value:.4f}")
print(f"  Significance level α  : {alpha}")
print(f"{'─'*46}")
print(f"  Decision              : {decision}")
print(f"{'─'*46}")
print(f"\n  Conclusion:")
print(f"  The clinic's mean glucose of {x_bar} mg/dL is significantly")
print(f"  higher than the national average of {mu_0} mg/dL")
print(f"  (Z = {Z:.2f}, p = {p_value:.4f}, α = {alpha})")

# ── Step 5: Visualise — the Z distribution with rejection region ──
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle('Z-Test Visualisation — Diabetes Clinic Glucose',
             fontsize=13, fontweight='bold')

x = np.linspace(-4, 4, 400)
y = stats.norm.pdf(x)

# Critical value for one-tailed test at alpha=0.05
z_critical = stats.norm.ppf(1 - alpha)   # = 1.645

# ── Left: Z distribution with rejection region shaded ─────────────
ax1 = axes[0]
ax1.plot(x, y, color='steelblue', linewidth=2.5)

# Shade rejection region (right tail beyond critical value)
x_reject = x[x >= z_critical]
ax1.fill_between(x_reject, stats.norm.pdf(x_reject),
                 color='tomato', alpha=0.5,
                 label=f'Rejection region (α=0.05)')

# Shade p-value area (right tail beyond our Z)
x_pval = x[x >= Z]
ax1.fill_between(x_pval, stats.norm.pdf(x_pval),
                 color='crimson', alpha=0.7,
                 label=f'p-value area = {p_value:.4f}')

ax1.axvline(Z, color='crimson', linewidth=2.5,
            linestyle='--', label=f'Our Z = {Z:.2f}')
ax1.axvline(z_critical, color='tomato', linewidth=2,
            linestyle=':',  label=f'Critical Z = {z_critical:.3f}')
ax1.axvline(0, color='steelblue', linewidth=1.5,
            linestyle='-',  alpha=0.5, label='H₀: Z = 0')

ax1.set_title('Standard Normal Distribution\nOne-Tailed Z-Test',
              fontweight='bold')
ax1.set_xlabel('Z-score')
ax1.set_ylabel('Probability Density')
ax1.legend(fontsize=8)

# ── Right: Original scale (mg/dL) ─────────────────────────────────
ax2 = axes[1]
x_raw = np.linspace(mu_0 - 4*SE, mu_0 + 4*SE, 400)
y_raw = stats.norm.pdf(x_raw, mu_0, SE)

ax2.plot(x_raw, y_raw, color='steelblue', linewidth=2.5)

# Rejection region on original scale
x_crit_raw = mu_0 + z_critical * SE
x_rej_raw  = x_raw[x_raw >= x_crit_raw]
ax2.fill_between(x_rej_raw, stats.norm.pdf(x_rej_raw, mu_0, SE),
                 color='tomato', alpha=0.5, label='Rejection region')

x_pv_raw = x_raw[x_raw >= x_bar]
ax2.fill_between(x_pv_raw, stats.norm.pdf(x_pv_raw, mu_0, SE),
                 color='crimson', alpha=0.7,
                 label=f'p-value = {p_value:.4f}')

ax2.axvline(x_bar,      color='crimson',   linewidth=2.5,
            linestyle='--', label=f'Sample mean = {x_bar} mg/dL')
ax2.axvline(x_crit_raw, color='tomato',    linewidth=2,
            linestyle=':',  label=f'Critical value = {x_crit_raw:.1f} mg/dL')
ax2.axvline(mu_0,       color='steelblue', linewidth=1.5,
            linestyle='-',  alpha=0.5, label=f'H₀ mean = {mu_0} mg/dL')

ax2.set_title('Original Scale (mg/dL)\nSame test — easier to interpret',
              fontweight='bold')
ax2.set_xlabel('Blood Glucose (mg/dL)')
ax2.set_ylabel('Probability Density')
ax2.legend(fontsize=8)

plt.tight_layout()
plt.savefig('z_test.png', dpi=150, bbox_inches='tight')
plt.show()
print("Chart saved as z_test.png")
```
