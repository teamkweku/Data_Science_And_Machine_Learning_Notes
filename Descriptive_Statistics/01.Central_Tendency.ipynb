````python
import pandas as pd
import numpy as np

# Create a sample healthcare dataset
data = {
    "age": [25, 32, 47, 51, 62, 23, 40, 36, 29, 55],
    "blood_pressure": [120, 130, 140, 135, 150, 110, 145, 138, 125, 155],
    "cholesterol": [180, 190, 210, 220, 240, 170, 230, 200, 185, 250],
    "bmi": [22.5, 24.0, 27.5, 30.0, 32.5, 21.0, 28.0, 26.5, 23.0, 31.0],
    "risk": [0, 0, 1, 1, 1, 0, 1, 0, 0, 1]  # 0 = low risk, 1 = high risk
}

df = pd.DataFrame(data)

df
````

````python
df.info()
````

## Measure of Central Tendency
- Testing the central tendency of this data points using python

````python
# Mean for each column
df.mean(numeric_only=True)

# Testing for a specific column
mean_blood_pressure = df["blood_pressure"].mean()
print(mean_blood_pressure)
````

````python
# Finding the median
print(df["blood_pressure"].median())
````

````python
# Fing the mode, the value that appears very often
print(df["blood_pressure"].mode())
````

## Outliers

````python
df_with_outliers = df.copy()
df_with_outliers.loc[len(df)] = [45, 300, 500, 40, 1]

df_with_outliers
````

## Mimicing a Practical Dataset to Learn the Measure of Dispersion of Data


````python
import pandas as pd
import numpy as np

np.random.seed(42)

n = 120

data = {
    "age": np.random.randint(18, 65, n),
    "monthly_income": np.random.randint(1000, 10000, n),
    "spending_score": np.random.randint(1, 100, n),
    "city": np.random.choice(["Accra", "Kumasi", "Tema", "Takoradi"], n),
    "subscription_type": np.random.choice(["Basic", "Standard", "Premium"], n)

}

df = pd.DataFrame(data)

df.head()
````

````python
df.info()
````

````python
df.describe()
````

## Introduction of Missing Values on Purpose

````python
# Introduce missing values randomly
for col in df.columns:
    df.loc[df.sample(frac=0.1).index, col] = np.nan

df.isnull().sum()
````

````python
df.info()
````

### Handling Missing Values
# Dropping the missing values
```python
df = df.dropna()
df.info()
```
# Imputing the missing values
```python
df["monthly_income"].fillna(df["monthly_income"].mean(), inplace=True)
```
- Numeric columns
Rule:
- Use *MEAN* -> If the data is *symmetric* and *normally distributed*
- Use *MEDIAN* -> If the data is *skewed*

- Use *MODE* -> If the data is *categorical*





````python
## Filling the dataset replacing null values with the mean since generated data seem symmetricl
df.isnull().sum()

df["age"] = df["age"].fillna(df["age"].mean())
df["monthly_income"] = df["monthly_income"].fillna(df["monthly_income"].mean())
df["spending_score"]=df["spending_score"].fillna(df["spending_score"].median())


df.isnull().sum()
````

````python
## Filling Categorical Columns with Mode
df["city"].mode()
df["subscription_type"].mode()

df["city"] = df["city"].fillna(df["city"].mode()[0])
df["subscription_type"] = df["subscription_type"].fillna(df["subscription_type"].mode()[0])

df.isnull().sum()
````

````python
df.describe(include="all")

df["monthly_income"].mean(), df["monthly_income"].median()
````

````python
## Exaggrating the situation for monthly income with outliers to see how it affects the mean & median
df_extreme = df.copy()
df_extreme.loc[0, "monthly_income"] = 100000 #extreme outlier

df_extreme["monthly_income"].mean(), df_extreme["monthly_income"].median()
````
